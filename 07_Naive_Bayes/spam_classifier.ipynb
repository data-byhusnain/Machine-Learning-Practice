```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
from sklearn.naive_bayes import GaussianNB,MultinomialNB,BernoulliNB

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,ConfusionMatrixDisplay
```

```python
# Sample Text Dataset (Spam vs Ham/Normal)
data = {
    'text': [
        'Win a free iphone now! Click here to claim your prize money',
        'Hey mum, are we still meeting for lunch today at 1pm?',
        'URGENT! You have won $1000 cash reward. Call this number now',
        'Can you send me the lecture notes for machine learning class?',
        'Free entry in a weekly competition to win prize tickets',
        'Let us play football this weekend in the evening',
        'Congratulations! You are selected for a free trip to Dubai',
        'Don us forget to bring your laptop for the meeting tomorrow'
    ],
    'label': ['spam', 'ham', 'spam', 'ham', 'spam', 'ham', 'spam', 'ham']
}
```

```python
df = pd.DataFrame(data)

df['target'] = df['label'].map({'ham':0,'spam':1})
print("dataset overview")
df[['text', 'target']]
```

```python
cv = CountVectorizer(stop_words='english')

X_text = cv.fit_transform(df['text']).toarray()
y = df['target'].values

print(X_text.shape)
print("Vocabulary (Pehele 10 words):", list(cv.vocabulary_.keys())[:10])
```

```python
mnb = MultinomialNB(alpha=1)
mnb.fit(X_text,y)

y_pred = mnb.predict(X_text)

print("\nTraining Accuracy:", accuracy_score(y, y_pred) * 100, "%")
print("\nClassification Report:")
print(classification_report(y, y_pred, target_names=['Ham (Normal)', 'Spam']))
```

```python
new_messages = [
    "Win $5000 prize now! Free gift card waiting for you",
    "Hey friend, let's meet tomorrow for movie",
    "i am john, and i am giving away free cashprizes"
]

# Naye messages ko same vectorizer se transform karein
new_X = cv.transform(new_messages).toarray()

# Prediction
predictions = mnb.predict(new_X)
probabilities = mnb.predict_proba(new_X)

for msg, pred, prob in zip(new_messages, predictions, probabilities):
    label = "SPAM 🚨" if pred == 1 else "HAM (Normal) ✅"
    confidence = np.max(prob) * 100
    print(f"\nMessage: '{msg}'")
    print(f"Prediction: {label} (Confidence: {confidence:.2f}%)")
```
